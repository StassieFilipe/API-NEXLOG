```python
import dask.dataframe as dd
import os

# 1. Definir o caminho correto dentro do Drive
# Se o seu arquivo está na pasta principal do Drive, o caminho é este:
caminho_drive = '/content/drive/MyDrive/API_2LOG/estoque_unificado_final.csv'

# 2. Verificar se o arquivo realmente está lá antes de tentar ler
if os.path.exists(caminho_drive):
    print("Arquivo encontrado! Iniciando processamento...")

    # Lendo com Dask
    # IMPORTANTE: mude 'unidade_medida' para o nome exato da sua coluna
    df = dd.read_csv(caminho_drive, dtype={'Unidade de Medida': 'str'})

    df['Unidade de Medida'] = 'litro'

    # Salvando o resultado no Drive
    caminho_saida = '/content/drive/MyDrive/API_2LOG/estoque_atualizado_litro.csv'
    df.to_csv(caminho_saida, index=False, single_file=True)

    print(f"Sucesso! Arquivo salvo em: {caminho_saida}")
else:
    print("ERRO: Arquivo não encontrado no caminho especificado.")
    print("Arquivos disponíveis no seu Drive (MyDrive):")
    # Isso vai listar os arquivos para você conferir o nome real
    print(os.listdir('/content/drive/MyDrive/'))
```
